# Revisão de Features

## Datas
<br>

#### Foi unido as datas e horários, alterado para o tipo data para fazer análise temporal e removido as que estavam no formato texto.

```python
# Combinar data + hora para criar uma nova coluna datetime completo
df_analysis['DataHora_primeira_observacao'] = pd.to_datetime(
    df_analysis['Data_da_primeira_observacao'].astype(str) + ' ' + df_analysis['Hora_da_primeira_observacao'].astype(str),
    errors='coerce'
)
df_analysis['DataHora_incidente'] = pd.to_datetime(
    df_analysis['Data_estimada_do_incidente'].astype(str) + ' ' + df_analysis['Hora_do_incidente'].astype(str),
    errors='coerce'
)

# Transformando a coluna para o tipo data
df_analysis['Data_de_criacao'] = pd.to_datetime(df_analysis['Data_de_criacao'], errors='coerce')

# Removendo as colunas que foram usadas para criar as novas colunas (pra não ficar repetido)
df_analysis = df_analysis.drop([
    'Data_da_primeira_observacao',
    'Hora_da_primeira_observacao',
    'Data_estimada_do_incidente',
    'Hora_do_incidente'], axis=1)
```

### Valor Minimo
<br>

#### Ao explorar um pouco os valores, percebi que a coluna DataHora_incidente tem um valor do ano de 1900 e outro de 1985, esses valores podem ser reais, seria interessante verificar junto a empresa que preencheu sobre a confiabilidade dessa informação, porém é muito estranho, pois após esses dois valores há um salto pro ano de 2011, então leva a crer que são valores errados e que podem impactar muito na análise temporal, então esses dois valores vão ser transformados em NaNs, entendendo que não são valores a serem levados em consideração.

#### Além disso tem uma única linha com ano de 2011, que vai impactar no momento da anáise, para ter uma ideia anual é necessário vários registros, então esse dado vai ser passado para 2012.

```python
# Alterando a única linha que é 2011 para 2012
df_analysis.loc[df_analysis['DataHora_incidente'].dt.year == 2011, 'DataHora_incidente'] = \
    df_analysis.loc[df_analysis['DataHora_incidente'].dt.year == 2011, 'DataHora_incidente'].apply(
        lambda x: x.replace(year=2012)
    )
```

```python
# Alterando essas linhas que estão com datas extremas para valores vazios
df_analysis.loc[22564, 'DataHora_incidente'] = pd.NaT
df_analysis.loc[14519, 'DataHora_incidente'] = pd.NaT
```

### Valores faltantes
<br>

#### Para os valores faltantes, vou usar as linhas que tem alguma data preenchida para copiar para a coluna que está com o dado faltante, assim é possível que se aproxime bem ao valor real e não deixa com o dado faltante.
<br>

#### A lógica vai ser: 
#### DataHora_primeira_observacao (COPIA) de Data_de_criacao
#### DataHora_incidente (COPIA) de DataHora_primeira_observacao
<br>

#### Escolhi essa lógica porque a DataHora_primeira_observacao tem 68 dados faltantes e a DataHora_incidente tem 3010, além disso a DataHora_incidente é mais próxima a primeira observação do que a de criação.

```python
# Substituindo os valores NaN de DataHora_primeira_observacao pelos valores de Data_de_criacao
df_analysis['DataHora_primeira_observacao'] = df_analysis['DataHora_primeira_observacao'].fillna(
    df_analysis['Data_de_criacao']
)

# Substituindo os valores NaN de DataHora_incidente pelos valores de DataHora_primeira_observacao
df_analysis['DataHora_incidente'] = df_analysis['DataHora_incidente'].fillna(
    df_analysis['DataHora_primeira_observacao']
)
```

## Empresa
<br>

#### Algumas empresas estão com o nome diferente, porém tem o mesmo código de identificação na frente, mostrando que são na realidade da mesma empresa. Isso pode significar que seja de um polo diferente ou pode ser um registro duplo, que pode fracionar as informações e levar uma análise equivocada.

#### Como a empresa pode estar com nome antigo, fantasia, ter mais de um nome, etc... Resolvi por deixar apenas o CNPJ e remover o nome da empresa, dessa forma esses dados podem ser agrupados a um único CNPJ.

#### As linhas mostradas abaixo não tem o nome ou CNPJ das empresas, achei esses valores após ter criado a coluna CNPJ e perceber que ficou com valores nulos (pois não tinham o CNPJ como padrão entre parenteses), mas todos tem como padrão o valor de 	"4.40237E+13" que é uma notação ciêntifica e tem suas instalações preenchidas, então da pra usar um documento fornecido pela ANP(Agência Nacional do Petróleo, Gás Natural e Biocombustíveis) para associar a instalação a empresa e depois verificar usando serasa experian.
<br>

#### Documento ANP: https://www.gov.br/anp/pt-br/assuntos/exploracao-e-producao-de-oleo-e-gas/desenvolvimento-e-producao/arquivos-lista-de-codigos/codigo-instalacoes.csv

#### Aqui foi feito um dicionário onde a chave é o valor da coluna Instalacao e o valor foi pego na pesquisa do documento disponibilizado pela ANP.
<br>

#### Logo depois é aplicado função que compara os valores da chave do dicionário com os valores da coluna Instalacao e muda o valor dessa linha na coluna Empresa pelo valor do dicionário. 

```python
# Criando valores para as linhas que não tem valor valido na coluna Empresa.
mapa_empresa = {
    "Unidade de Compressão de Gás Natural do Campo de Graúna": 
        "PHOENIX ÓLEO & GÁS NATURAL LTDA. (32528443000146)",

    "Estação Coletora João de Barro": 
        "NÍON ENERGIA S.A. (45832752000157)",

    "Estação coletora Periquito": 
        "PHOENIX EMPREENDIMENTOS LTDA. (05282436000134)",

    "Estação Coletora Concriz": 
        "PHOENIX EMPREENDIMENTOS LTDA. (05282436000134)",

    "Campo Galo de Campina - Poço 3-UTC-5A-RN": 
        "NÍON ENERGIA S.A. (45832752000157)",

    "Estação Coletora Galo de Campina": 
        "NÍON ENERGIA S.A. (45832752000157)",

    "Campo João de Barro - Poço 4-AURI-15-RN": 
        "NÍON ENERGIA S.A. (45832752000157)",

    "3-AURI-09-RN (Campo João de Barro)": 
        "NÍON ENERGIA S.A. (45832752000157)",
}

# Preenche os valores da coluna Empresa de acordo com o mapa_empresa criado antes que tem como chave a Instalação
df_analysis['Empresa'] = df_analysis.apply(
    lambda row: mapa_empresa.get(row['Instalacao'], row['Empresa']),
    axis=1
)
```

```python
# Criando a coluna CNPJ que vai conter apenas o código que está entre parenteses
df_analysis['CNPJ'] = df_analysis['Empresa'].str.extract(r'\((\d+)\)')

# Verificando quantas empresas tem valores repetidos
df_analysis.groupby('CNPJ')['Empresa'].nunique().sort_values(ascending=False)
```

```python
# Removendo a coluna "Empresa", pois já tem a nova coluna CNPJ para identifica-las
df_analysis.drop(columns=['Empresa'], inplace=True)
```

## Numero_de_feridos_graves
<br>

#### Essa coluna tem alguns valores que não tem haver com a quantidade de feridos graves, como data/hora, NaN(valor vazio), e espaço em branco ' '. Então basta retirar esses valores que não tem haver e preencher o NaN com 0 que é a moda, sendo que tem apenas 0.24% de valores NaN, dessa forma vai ficar tudo certo com a coluna e não vai ter nenhum impacto por esse preenchimento.

```python
# Verificando cada tipo de valor que tem na coluna e suas respectivas quantidades
df_analysis['Numero_de_feridos_graves'].value_counts()
```

```python
# Transforma tudo que for possível para numérico e o que da erro vai virar NaN
df_analysis['Numero_de_feridos_graves'] = pd.to_numeric(
    df_analysis['Numero_de_feridos_graves'], errors='coerce'
)

# Transforma tudo que é NaN em 0 que é a moda com uma quantidade muito maior
df_analysis['Numero_de_feridos_graves'] = (
    df_analysis['Numero_de_feridos_graves']
    .fillna(0)
    .astype(int)
)
```

## Tipo_de_incidente
<br>

####  Essa coluna está bem padronizada, ainda assim tem muitos valores diferentes, sendo 186 tipos, não da pra ter certeza se foi usado um tipo de seleção restrita para o usuário ou apenas foi bem padronizado por responsáveis da inclusão dessa informação. Então para focar apenas em informações mais confiaveis, foi deixado apenas os preenchimentos com SSM na frente, preenchidos pela Segurança, Saúde e Meio Ambiente, o restante foram preenchidos como "Informação não padronizada".

```python
df_analysis['Tipo_de_incidente'].str.startswith('SSM').sum()
```

```python
# Alterando os valores NaN e outros aleatórios por "Informação não padronizada"
df_analysis['Tipo_de_incidente'] = (
    df_analysis['Tipo_de_incidente']
    .fillna("Informação não padronizada")
    .where(lambda s: s.str.startswith("SSM"), "Informação não padronizada")
)
```

## Numero_de_fatalidades
<br>

#### Muito similar ao feridos graves, algumas informações que não tem sentido com o que a coluna está pedindo, então será usado a mesma estratégia, transformar o que é número em número, o resto vira NaN e depois os NaN vão virar a moda.

```python
# Verificando cada tipo de valor que tem na coluna e suas respectivas quantidades
df_analysis['Numero_de_fatalidades'].value_counts()
```

```python
# Transforma tudo que for possível para numérico e o que da erro vai virar NaN
df_analysis['Numero_de_fatalidades'] = pd.to_numeric(
    df_analysis['Numero_de_fatalidades'], errors='coerce'
)

# Transforma tudo que é NaN em 0 que é a moda com uma quantidade muito maior
df_analysis['Numero_de_fatalidades'] = (
    df_analysis['Numero_de_fatalidades']
    .fillna(0)
    .astype(int)
)
```

## DSC_GRAVIDADE_TIPO
<br>

#### Novamente um problema bem similar onde a coluna tem alguns valores inválidos, sendo os únicos validos LEVE, MODERADO E GRAVE, então os demais valores vão ser transformados na moda novamente.

```python
# Verificando cada tipo de valor que tem na coluna e suas respectivas quantidades
df_analysis['DSC_GRAVIDADE_TIPO'].value_counts()
```

```python
# Atribuindo a moda a variável moda
moda = df_analysis['DSC_GRAVIDADE_TIPO'].mode()[0]

# Alterando todos valores inadequados com a moda
df_analysis['DSC_GRAVIDADE_TIPO'] = df_analysis['DSC_GRAVIDADE_TIPO'].apply(
    lambda x: x if x in ['LEVE', 'MODERADO', 'GRAVE'] else moda
)
```

## DSC_QUASE_ACIDENTE_ACIDENTE
<br>

#### A coluna de quase acidente também infrenta o problema de alguns valores preenchidos de forma errada, mas ao verificar a quantidade de cada valor da pra perceber que deveria ser preenchido um "Quase acidente" ou de fato um "Acidente", os demais valores parecem não ser validos, então serão transformados em moda novamente.


```python
# Verificando cada tipo de valor que tem na coluna e suas respectivas quantidades
df_analysis['DSC_QUASE_ACIDENTE_ACIDENTE'].value_counts()
```

```python
# Atribuindo a moda a variável moda
moda = df_analysis['DSC_QUASE_ACIDENTE_ACIDENTE'].mode()[0]

# Alterando todos valores inadequados com a moda
df_analysis['DSC_QUASE_ACIDENTE_ACIDENTE'] = df_analysis['DSC_QUASE_ACIDENTE_ACIDENTE'].apply(
    lambda x: x if x in ['Quase acidente', 'Acidente'] else moda
)
```

## Tipo_Instalacao_Detalhado
<br>

#### A coluna Tipo_Instalacao_Detalhado é um pouco similar as colunas onde tinham alguns valores inadequados e foram preenchidos pela moda, aqui também tiveram valores incorretos, porém para alterar esses valores foi necessário olhar valor por valor e ver qual tinha alguma relação com a proposta da coluna e qual foi preenchido completamente errado, então fiz a alteração de forma bem arbitraria e ao invés de substituir pela moda que é FPSO, substitui pelo valor "Outro", já que não tem como eu saber ao certo o tipo de instalação e dessa forma eu deixo em aberto o que poderia ter sido.

```python
# Verificando cada tipo de valor que tem na coluna e suas respectivas quantidades
df_analysis['Tipo_Instalacao_Detalhado'].value_counts()
```

```python
# Lista de valores inválidos que devem virar "Outro"
valores_invalidos = [
    '2/1/2022', '0', '8/2/2021',
    'Falha de elemento do conjunto solidário de barreira (CSB)',
    '1/18/2022', '13:10', '13:14', '10:20', '15:05',
    '14:00', '9:10', '9:45', 'Acidente'
]

# Substitui os valores invalidos e os NaN por "Outro"
df_analysis['Tipo_Instalacao_Detalhado'] = (
    df_analysis['Tipo_Instalacao_Detalhado']
    .replace(valores_invalidos, "Outro")
    .fillna("Outro")
)
```

## Custo_Incidente
<br>

#### Criação da coluna Custo_Incidente para inserir valores numéricos para comparação dos incidentes.

```python
# Dicionários de gravidades para cada categoria
gravidade_downtime = {
    "LEVE": 1,
    "MODERADO": 9,
    "GRAVE": 30
}

custos_ambientais = {
    "LEVE": 3000.00,
    "MODERADO": 252500.00,
    "GRAVE": 25025000.00
}

custos_saude = {
    "LEVE": 2250.00,
    "MODERADO": 249750.00,
    "GRAVE": 25025000.00
}

custos_equip = {
    "LEVE": 45000.00,
    "MODERADO": 2250000.00,
    "GRAVE": 22500000.00
}

custos_perda_controle = {
    "LEVE": 225000.00,
    "MODERADO": 2250000.00,
    "GRAVE": 22500000.00
}

def calcular_custo(row):
    categoria = row["Categoria_Custo"]
    grav = row["DSC_GRAVIDADE_TIPO"]
    feridos_graves = row["Numero_de_feridos_graves"]

    # Downtime
    if categoria == "Downtime":
        return 149710.00 * gravidade_downtime[grav]

    # Custos Ambientais
    elif categoria == "Custos Ambientais":
        return custos_ambientais[grav]

    # Danos à Saúde (com multiplicação por número de feridos graves)
    elif categoria == "Danos a Saúde":
        custo_base = custos_saude[grav]

        if feridos_graves >= 2:
            return custo_base * feridos_graves
        else:
            return custo_base

    # Danos a Equipamentos
    elif categoria == "Danos a Equipamentos":
        return custos_equip[grav]

    # Perda de Controle
    elif categoria == "Perda de Controle/Critico":
        return custos_perda_controle[grav]

    # Não Classificado
    elif categoria == "Não Classificado":
        return 1

    else:
        return np.nan  # caso apareça alguma categoria inesperada


# Criar a coluna
df_analysis["Custo_Incidente"] = df_analysis.apply(calcular_custo, axis=1)
```
